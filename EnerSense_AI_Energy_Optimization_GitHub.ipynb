# ⚡ EnerSense AI — Intelligent Building Energy Optimizer

## 1. Project Overview

EnerSense AI is a machine-learning prototype for short-term building energy forecasting and operational alerting. It combines historical appliance consumption with environmental and time-based features to estimate near-term consumption, identify unusually high demand, and generate simple operational recommendations.

````python
# Split the data chronologically into 80% training data and 20% testing data

split_index = int(len(df) * 0.8)

X_train = X.iloc[:split_index]
X_test = X.iloc[split_index:]

y_train = y.iloc[:split_index]
y_test = y.iloc[split_index:]

print("X_train:", X_train.shape)
print("X_test :", X_test.shape)
print("y_train:", y_train.shape)
print("y_test :", y_test.shape)
````

````python
# Train a Linear Regression model as our baseline model

from sklearn.linear_model import LinearRegression

model_lr = LinearRegression()

model_lr.fit(X_train, y_train)

print("Linear Regression model trained successfully.")
````

````python
# Use the trained model to predict energy consumption for the test data

y_pred_lr = model_lr.predict(X_test)

print("First 10 actual values:")
print(y_test.head(10).values)

print("\nFirst 10 predicted values:")
print(y_pred_lr[:10])
````

````python
# Evaluate the Linear Regression model using standard regression metrics

from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

mae_lr = mean_absolute_error(y_test, y_pred_lr)
rmse_lr = np.sqrt(mean_squared_error(y_test, y_pred_lr))
r2_lr = r2_score(y_test, y_pred_lr)

print("Linear Regression Performance")
print("MAE :", mae_lr)
print("RMSE:", rmse_lr)
print("R²  :", r2_lr)
````

````python
# Create the feature set with 10-, 20-, and 30-minute historical energy values

feature_columns_lags = [
    "lights",
    "T1", "RH_1",
    "T2", "RH_2",
    "T3", "RH_3",
    "T4", "RH_4",
    "T5", "RH_5",
    "T6", "RH_6",
    "T7", "RH_7",
    "T8", "RH_8",
    "T9", "RH_9",
    "T_out",
    "Press_mm_hg",
    "RH_out",
    "Windspeed",
    "Visibility",
    "Tdewpoint",
    "hour",
    "day_of_week",
    "is_weekend",
    "month",
    "Appliances_lag_1",
    "Appliances_lag_2",
    "Appliances_lag_3"
]

X_lags = df_model_lags[feature_columns_lags]
y_lags = df_model_lags["Appliances"]

print("Features:", X_lags.shape)
print("Target:", y_lags.shape)
print("Last 3 features:", X_lags.columns[-3:].tolist())
````

````python
# Split the 3-lag dataset chronologically into 80% training and 20% testing data

split_index_lags = int(len(X_lags) * 0.8)

X_train_lags = X_lags.iloc[:split_index_lags]
X_test_lags = X_lags.iloc[split_index_lags:]

y_train_lags = y_lags.iloc[:split_index_lags]
y_test_lags = y_lags.iloc[split_index_lags:]

print("X_train_lags:", X_train_lags.shape)
print("X_test_lags :", X_test_lags.shape)
print("y_train_lags:", y_train_lags.shape)
print("y_test_lags :", y_test_lags.shape)
````

````python
# Train and evaluate Linear Regression using 10-, 20-, and 30-minute lag features

model_lr_lags = LinearRegression()
model_lr_lags.fit(X_train_lags, y_train_lags)

y_pred_lr_lags = model_lr_lags.predict(X_test_lags)

mae_lr_lags = mean_absolute_error(y_test_lags, y_pred_lr_lags)
rmse_lr_lags = np.sqrt(mean_squared_error(y_test_lags, y_pred_lr_lags))
r2_lr_lags = r2_score(y_test_lags, y_pred_lr_lags)

print("3-Lag Linear Regression Performance")
print("MAE :", mae_lr_lags)
print("RMSE:", rmse_lr_lags)
print("R²  :", r2_lr_lags)
````

````python
high_threshold = 196

high_actual = y_test_lags >= high_threshold
high_predicted = y_pred_lr_lags >= high_threshold

print("Actual high-consumption events:", high_actual.sum())
print("Predicted high-consumption events:", high_predicted.sum())
````

````python
from sklearn.metrics import confusion_matrix

cm = confusion_matrix(high_actual, high_predicted)

print("Confusion Matrix:")
print(cm)

tn, fp, fn, tp = cm.ravel()

print("True Negatives :", tn)
print("False Positives:", fp)
print("False Negatives:", fn)
print("True Positives :", tp)
````

````python
from sklearn.metrics import precision_score, recall_score

# Precision:
# Of all the high-consumption alerts we generated,
# how many were actually high-consumption events?
precision = precision_score(high_actual, high_predicted)

# Recall:
# Of all the actual high-consumption events,
# how many did our model successfully detect?
recall = recall_score(high_actual, high_predicted)

print("Peak Alert Performance")
print("Precision:", precision)
print("Recall   :", recall)
````

````python
# Define the two data-driven energy thresholds
high_threshold = 196      # 90th percentile
very_high_threshold = 330 # 95th percentile

# Create an alert level from the model's predicted consumption
def get_alert_level(predicted_energy):
    if predicted_energy >= very_high_threshold:
        return "VERY HIGH"
    elif predicted_energy >= high_threshold:
        return "HIGH"
    else:
        return "NORMAL"

# Apply the alert logic to every prediction
alert_levels = [get_alert_level(value) for value in y_pred_lr_lags]

# Display how many predictions fall into each alert category
print(pd.Series(alert_levels).value_counts())
````

````python
# Find the test observations where our model predicted VERY HIGH consumption
very_high_mask = y_pred_lr_lags >= very_high_threshold

# Get the corresponding rows from the test dataset
very_high_cases = X_test_lags.loc[very_high_mask].copy()

# Add the actual and predicted energy values
very_high_cases["Actual_Energy"] = y_test_lags.loc[very_high_mask].values
very_high_cases["Predicted_Energy"] = y_pred_lr_lags[very_high_mask]

# Display the first 5 very-high cases
print(very_high_cases[
    ["hour", "T_out", "RH_out", "lights",
     "Appliances_lag_1", "Appliances_lag_2", "Appliances_lag_3",
     "Actual_Energy", "Predicted_Energy"]
].head())
````

````python
# Create recommendations for VERY HIGH predicted consumption
recommendations = []

for prediction, lag_1 in zip(y_pred_lr_lags, X_test_lags["Appliances_lag_1"]):

    # VERY HIGH prediction + high recent consumption
    if prediction >= very_high_threshold and lag_1 >= high_threshold:
        recommendation = "Check major electrical loads"

    # VERY HIGH prediction but recent consumption is lower
    elif prediction >= very_high_threshold:
        recommendation = "Monitor building energy usage"

    # No elevated prediction
    else:
        recommendation = "Normal operation"

    recommendations.append(recommendation)

# Count each recommendation
print(pd.Series(recommendations).value_counts())
````

````python
# Select the test observations where the model predicted VERY HIGH consumption
very_high_cases = X_test_lags[y_pred_lr_lags >= very_high_threshold].copy()

# Calculate the average conditions during these VERY HIGH predictions
print("Average conditions during VERY HIGH predictions:")
print(very_high_cases[
    ["hour", "T_out", "RH_out", "lights"]
].mean())
````

````python
# Create a summary table for the recommendation engine
recommendation_summary = pd.Series(recommendations).value_counts().reset_index()

# Rename the columns for a cleaner dashboard-ready format
recommendation_summary.columns = ["Recommendation", "Count"]

# Calculate the percentage of test observations for each recommendation
recommendation_summary["Percentage"] = (
    recommendation_summary["Count"] / len(recommendations) * 100
)

print(recommendation_summary)
````

````python
# Calculate the prediction error for every test observation
# Positive value = actual consumption was higher than predicted
# Negative value = actual consumption was lower than predicted
residuals = y_test_lags.values - y_pred_lr_lags

# Convert residuals into absolute errors
# This tells us how far the prediction was from reality,
# regardless of whether it was too high or too low.
absolute_errors = np.abs(residuals)

# Display the distribution of prediction errors
print(pd.Series(absolute_errors).describe(
    percentiles=[0.75, 0.90, 0.95, 0.99]
))
````

````python
# Use the 95th percentile of prediction error as the anomaly threshold
anomaly_threshold = np.percentile(absolute_errors, 95)

# Flag observations where actual consumption differs
# substantially from the model's prediction
anomaly_flags = absolute_errors >= anomaly_threshold

# Count normal and anomalous observations
print("Anomaly threshold:", anomaly_threshold)
print("Anomalies detected:", anomaly_flags.sum())
print("Anomaly percentage:", anomaly_flags.mean() * 100)
````

````python
# Select the test observations flagged as anomalies
anomaly_cases = X_test_lags[anomaly_flags].copy()

# Add the actual energy, predicted energy, and prediction error
anomaly_cases["Actual_Energy"] = y_test_lags[anomaly_flags].values
anomaly_cases["Predicted_Energy"] = y_pred_lr_lags[anomaly_flags]
anomaly_cases["Absolute_Error"] = absolute_errors[anomaly_flags]

# Sort by the largest prediction error
anomaly_cases = anomaly_cases.sort_values(
    "Absolute_Error",
    ascending=False
)

# Display the 10 most significant anomalies
print(anomaly_cases[
    ["hour", "T_out", "RH_out", "lights",
     "Appliances_lag_1",
     "Actual_Energy",
     "Predicted_Energy",
     "Absolute_Error"]
].head(10))
````

````python
# Calculate the signed prediction error
# Positive value = actual consumption was higher than predicted
# Negative value = actual consumption was lower than predicted
signed_errors = y_test_lags.values - y_pred_lr_lags

# Flag only unusually HIGH actual consumption
# using our 95th-percentile anomaly threshold
high_anomaly_flags = signed_errors >= anomaly_threshold

print("Unexpected high-consumption events:", high_anomaly_flags.sum())
print(
    "Percentage of test observations:",
    high_anomaly_flags.mean() * 100
)
````

````python
# Identify events that are both:
# 1. Unexpectedly high compared with the prediction
# 2. Actually above our 196 Wh high-consumption threshold

combined_alerts = (
    high_anomaly_flags &
    (y_test_lags.values >= high_threshold)
)

print("Unexpected + high-consumption events:", combined_alerts.sum())
print(
    "Percentage of test observations:",
    combined_alerts.mean() * 100
)
````

````python
# Create a copy of the test data so we can attach our ML results
results = X_test_lags.copy()

# Add the actual energy consumption
results["Actual_Energy"] = y_test_lags.values

# Add the model's predicted energy consumption
results["Predicted_Energy"] = y_pred_lr_lags

# Add the prediction error
results["Prediction_Error"] = (
    results["Actual_Energy"] - results["Predicted_Energy"]
)

# Add the absolute prediction error
results["Absolute_Error"] = abs(results["Prediction_Error"])

# Create the energy alert level
results["Alert_Level"] = [
    get_alert_level(prediction)
    for prediction in results["Predicted_Energy"]
]

# Add the unexpected high-consumption anomaly flag
results["Anomaly"] = high_anomaly_flags

# Add the recommendation generated earlier
results["Recommendation"] = recommendations

# Display the first 5 rows of the final results
print(results[
    [
        "hour",
        "Actual_Energy",
        "Predicted_Energy",
        "Alert_Level",
        "Anomaly",
        "Recommendation"
    ]
].head())
````

````python
# Create the improved recommendation logic
final_recommendations = []

for alert, anomaly in zip(
    results["Alert_Level"],
    results["Anomaly"]
):
    # Unexpected high consumption gets the highest priority
    if anomaly:
        recommendation = "Investigate unexpected energy increase"

    # Very high predicted consumption needs preventive action
    elif alert == "VERY HIGH":
        recommendation = "Check major electrical loads"

    # High but not anomalous
    elif alert == "HIGH":
        recommendation = "Monitor building energy usage"

    # Normal operation
    else:
        recommendation = "Normal operation"

    final_recommendations.append(recommendation)

# Replace the old recommendation with the improved one
results["Recommendation"] = final_recommendations

# Show the recommendation distribution
print(results["Recommendation"].value_counts())
````

````python
# Create one function containing our final decision logic
def energy_decision(predicted_energy, is_anomaly):

    # Very high predicted consumption
    if predicted_energy >= very_high_threshold:
        alert = "VERY HIGH"
        recommendation = "Check major electrical loads"

    # High predicted consumption
    elif predicted_energy >= high_threshold:
        alert = "HIGH"
        recommendation = "Monitor building energy usage"

    # Normal predicted consumption
    else:
        alert = "NORMAL"
        recommendation = "Normal operation"

    # Anomaly takes priority over the normal recommendation
    if is_anomaly:
        recommendation = "Investigate unexpected energy increase"

    return alert, recommendation


# Test the function with one example
test_alert, test_recommendation = energy_decision(
    predicted_energy=400,
    is_anomaly=False
)

print("Alert:", test_alert)
print("Recommendation:", test_recommendation)
````

````python
# Apply the final decision function to every test prediction
final_decisions = [
    energy_decision(prediction, anomaly)
    for prediction, anomaly in zip(
        results["Predicted_Energy"],
        results["Anomaly"]
    )
]

# Store the alert levels and recommendations separately
results["Final_Alert"] = [decision[0] for decision in final_decisions]
results["Final_Recommendation"] = [decision[1] for decision in final_decisions]

# Display the first 5 final decisions
print(results[
    [
        "Actual_Energy",
        "Predicted_Energy",
        "Final_Alert",
        "Anomaly",
        "Final_Recommendation"
    ]
].head())
````

## 🏆 Model Selection

We compare the models using the **same chronological test period**. The naive baseline is important because this is a time-dependent prediction problem: a complex model should beat a simple "use the previous observation" strategy before it is considered useful.


````python
model_comparison = pd.DataFrame({
    "Model": [
        "Naive Previous Value",
        "Linear Regression",
        "Random Forest",
        "Gradient Boosting",
        "Linear Regression + 1 Lag",
        "Linear Regression + 3 Lags"
    ],
    "MAE (Wh)": [
        mae_naive, mae_lr, mae_rf, mae_gb, mae_lr_lag, mae_lr_lags
    ],
    "RMSE (Wh)": [
        rmse_naive, rmse_lr, rmse_rf, rmse_gb, rmse_lr_lag, rmse_lr_lags
    ],
    "R²": [
        r2_naive, r2_lr, r2_rf, r2_gb, r2_lr_lag, r2_lr_lags
    ]
}).sort_values("MAE (Wh)")

print(model_comparison.to_string(index=False))
print("\nSelected model: Linear Regression + 3 historical lag features")
````

### Interpretation

- The **3-lag Linear Regression** is the selected model because it has the strongest observed R² and lower error than the naive baseline in this notebook.
- Random Forest and Gradient Boosting are retained as comparison models; their negative R² on this chronological split indicates they do not generalize well here.
- The anomaly detector is **model-residual based**. It should be described as an "unexpected consumption detector", not as proof of equipment failure.
- Thresholds are data-driven: `HIGH` uses the 90th percentile of observed consumption and `VERY HIGH` uses the 95th percentile.
